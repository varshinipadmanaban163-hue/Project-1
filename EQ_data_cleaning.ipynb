# Data Preparation

```python
import pandas as pd 
import numpy as np
```

### Loading data in pandas

```python

df=pd.read_csv("project_eq_raw.csv")
```

### Converting time into datetime format

```python

df["time"]=pd.to_datetime(df["time"])
```

### Converting updated into datetime format

```python

df["updated"]=pd.to_datetime(df["updated"])
```

## Cleaning text fields

### Extract country using regex

```python

df["country"]=df['place'].str.extract(r',\s*([^,]+)$')
```

### Removing extra space and lowercase in string field

```python

str_field=['id','magType','place','status','net','types','ids','sources','type','alert']
```

```python

for i in str_field:
    df[i]=df[i].str.strip().str.lower()
```

```python
df['country'].str.title()
```

### Removing unwanted commas from  and types

```python
comma_columns = ["ids","types","sources"]
```

```python
for i in comma_columns:
    df[i]=df[i].str.replace(",","")
```

```python
df[comma_columns].head()
```

```python
df.duplicated("id").sum()
```

## Clean Numeric field

### Filling missing values with 0

```python

df=df.fillna(0)
```

```python
df.isnull()
```

## Add Derived Columns

### Adding new columns based on time

```python

df["year"]=df['time'].dt.year
df["month"]=df['time'].dt.month
df["day"]=df['time'].dt.day
df["day-of_week"]=df["time"].dt.day_of_week
```
